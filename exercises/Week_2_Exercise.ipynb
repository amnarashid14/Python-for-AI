````python
# Your practice space — question 1

# Write a function `top_items(n, *sales)`. Each sale is a `(item, amount)` tuple, and you can pass in as many as you like. 
# Return a list of the **names** of the `n` best-selling items, highest first. Use `sorted()` with a `lambda`.

# Test your code with:
# ```python
# print(top_items(2, ("pen", 50), ("book", 300), ("bag", 120)))
# print(top_items(5, ("pen", 50), ("book", 300)))
# print(top_items(3))
# ```
# Expected output:
# ```
# ['book', 'bag']
# ['book', 'pen']
# []
#n is the number of sales
def top_items(n, *sales):
    if n >= 1 or n <= sales[1]:
        sort = sorted(sales, key = lambda x: x[1] , reverse = True)
        # print(sort)
        for name in sort:
            return (name[0])
    else:
        print("Invalid Range!")

print("First", top_items(2, ("pen", 50), ("book", 300), ("bag", 120)))
print("Second" ,top_items(5, ("pen", 50), ("book", 300)))
print("Third",top_items(3))
````

````python
# Your practice space — question 2
# Write a class `Product` so that `Product("pen", 50, 10)` has a `name`, a `price` and a `stock` (stock defaults to 
# `0`). Give it three methods:

# - `restock(qty)` adds `qty` to the stock.
# - `sell(qty)` takes `qty` away and returns `True`. If there isn't enough stock, it changes **nothing** and
#  returns `False`.
# - `value()` returns `price × stock`.


class Product:
    def __init__(self, name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock

    def restock(self, qty):
        self.qty = qty
        print(f"Initial Stock: {self.stock}\nInitial Quantity :{qty} ")
        self.stock += qty
        print(f"After Stock: {self.stock} ")
        print("Inside Restock",qty)

    def sells(self, num_of_sales):
        if num_of_sales >= self.qty :
            return "Stock exceeded"
        else:
            self.qty -= num_of_sales  
            return f"Remaing Quantity : {self.qty}"


pr1 = Product ("abc", 123, 45)
pr1.restock(2)
pr1.sells(23)
````
