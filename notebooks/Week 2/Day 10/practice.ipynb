```python
# **1.** Write a `Student` class. `__init__` takes a `name` and a list of `marks`. Add a method `average()` that returns the mean mark, and a method `grade()` that returns `"A"` for 80+, `"B"` for 70+, `"C"` for 60+, else `"F"`. Test it on `Student("Fatima", [85, 92, 78])`.
```

```python
class Student:
    def __init__(self,name, marks):
        self.name = name
        self.marks = marks
        self.avg = 0
    def average(self):
        Sum= sum(self.marks)
        self.avg = Sum / len (self.marks)
        return f"The Average Marks of {self.name} is {self.avg}"
    def grade(self):
            if self.avg >= 80:
                return f"Grade : A"
            elif self.avg >= 70:
                return f"Grade : B"
            elif self.avg >= 60:
                return f"Grade : C"
            elif self.avg >= 50:
                return f"Grade : D"
            else:
                return f"Grade : Fail"

a = Student("abc",[80,96,53,78,55])
b = Student("Fatima",[85,92,78])
print(a.average())
print(a.grade())

print(b.average())
print(b.grade())
```
