```python
def count_up_to(n):
    i = 1
    while i <= n:
        yield i
        i += 1



print("For Loop Value..........")
for value in count_up_to(10):
    # print(value )
    # if value <= 100:
    #     gen_func.close()
    print("Gen", value)
# print(next(gen_func)) #1
# print(next(gen_func)) #8
# print(next(gen_func)) #15

```
