```python
# Write `safe_divide(a, b)` that raises a custom `DivisionByZeroError` (inherit from `Exception`) when `b` is `0`, and returns `a / b` otherwise.
#  Call it inside `try/except` with `b = 0` and print a friendly message. Add a `finally` that prints `"done"`.



def safe_divide(a,b):
    try:
        div = a/ b
        return div
    except Exception as e:
        return f"Error raised : {e}"
    finally:
        print("Done!")

print(safe_divide(2,0))
```
