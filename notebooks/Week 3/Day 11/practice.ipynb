```python
# 1.Build a shape family. `Shape` is the base with an `area()` that returns `0`.
# `Rectangle(width, height)` and `Circle(radius)` each override `area()`. 
# Put a few in a list and print each area in one loop (polymorphism).

class Shape:
    def __init__(self,name,width,height):
    # def __init__(self,width,height):
        self.name = name
        self.height = height
        self.width =width
    def area(self, area):
        area = 0
        return area

class Rectangle(Shape):
    def __init__(self, name,  width, height ):
        super().__init__(name, width,height)
        self.name = name

    def area(self):
        area = self.height * self.width
        print(f"The area of {self.name} is {area} ")

rec = Rectangle("rectangle",3,2)
rec.area()

class Circle(Shape):
    def __init__(self,name ,radius):
        self.radius = radius
        self.name = name

        

    def area(self):
        area = f"The area of {self.name} is {3.142* self.radius**2:.2f}"
        print(area)
         

circle = Circle("circle", 2)
circle.area()






    
```

```python
# 2. The classic: `Animal` with a `speak()`, then `Dog` and `Cat` overriding it.
# Loop over a mixed list and print what each says.


class Animal:
    def __init__(self,name):
        self.name = name

    def speak(self):
        return f"{self.name} sounds - "

class Dog(Animal):
    def speak(self):
        return " Woof!"


class Cat(Animal):
    def speak(self):
        return "Meow!"

# class Cat(Animal):
#     def speak(self):
#         base = super().speak()
#         return f"{base} Meow"

mano = Cat ("mano")
print(mano.speak())

for animal in [Dog("Tommy"), Cat("Kitty"), Animal("Xyz")]:
    print(f"{animal.name} says {animal.speak()}")
```

```python
# 3.Extend Class 10's account: write `SavingsAccount` and `CurrentAccount` as 
# children of `Account`. `SavingsAccount` overrides nothing but adds 
# `add_interest(rate)`; `CurrentAccount` overrides `withdraw` to allow an overdraft.
# Use `super().__init__(...)` where you add attributes.

class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self,amount):
        self.balance += amount
        return f"{self.owner} Deposited : {amount}"

    def withdraw(self,amount):
        if amount <= self.balance:
            self.balance -= amount
            return f"{self.owner} Withdrew : {amount}"
        else:
            return "Insufficient Balance - Cannot procceed further"

    def transfer(self,amount,other_acc):
        if amount <= self.balance:
            self.withdraw(amount)
            other_acc.deposit(amount)
            return f"An amount of Rs.{amount} has been transferred to {other_acc.owner} from {self.owner}\nYour Current Balance : {self.balance}"
        else:
            return f"Insufficient Balance!"


#first User
zara = BankAccount("Zara", 8000)
print("Zara's Balance", zara.balance)
print(zara.withdraw(10000))
print( zara.deposit(100))
print("Zara's Balance", zara.balance)
print("\n")

#second User
ali = BankAccount("Ali", 5000)
print("Ali's Balance", ali.balance)
print(ali.withdraw(10000))
print( ali.deposit(100))
print("Ali's Balance", ali.balance)
print("\n")
# print("Balance of Ali", ali.balance)
# print("Balance of Zara", zara.balance)


print(ali.transfer(10000, zara))
print(ali.transfer(100, zara))

print("\n")
print("New Balance of Zara", zara.balance)

```

```python
class Teacher:
    def __init__(self, name):
        self.name = name

    def teach(self):
        print(f"{self.name} is teaching.")


class MathTeacher(Teacher):
    def teach_a_lesson(self):                   # overriding the parent's version
        print(f"{self.name} is solving equations on the board.")


t1 = MathTeacher("ABC")
t1.teach()
t1.teach_a_lesson()

t2 = Teacher("ABC")
t2.teach()
```
